```python
def get_leverkusen_result(row):
    if row["home_team"] == "Bayer Leverkusen":
        leverkusen_score = row["home_score"]
        opponent_score = row["away_score"]
    else:
        leverkusen_score = row["away_score"]
        opponent_score = row["home_score"]

    if leverkusen_score > opponent_score:
        return "Win"
    elif leverkusen_score < opponent_score:
        return "Loss"
    else:
        return "Draw"
```

```python
match_summary_df["result"] = match_summary_df.apply(
    get_leverkusen_result,
    axis=1
)
```

```python
match_summary_df["opponent"] = match_summary_df.apply(
    lambda row: (
        row["away_team"]
        if row["home_team"] == "Bayer Leverkusen"
        else row["home_team"]
    ),
    axis=1,
)
```

```python
match_summary_df["venue"] = match_summary_df["home_team"].apply(
    lambda team: "Home" if team == "Bayer Leverkusen" else "Away"
)
```

```python
match_summary_df["goals_for"] = match_summary_df.apply(
    lambda row: (
        row["home_score"]
        if row["home_team"] == "Bayer Leverkusen"
        else row["away_score"]
    ),
    axis=1,
)

match_summary_df["goals_against"] = match_summary_df.apply(
    lambda row: (
        row["away_score"]
        if row["home_team"] == "Bayer Leverkusen"
        else row["home_score"]
    ),
    axis=1,
)
```

```python
match_summary_df["goal_difference"] = (
    match_summary_df["goals_for"]
    - match_summary_df["goals_against"]
)
```

# Data Audit Summary

The Bayer Leverkusen 2023/24 Bundesliga StatsBomb Open Data release contains:

- 34 Bundesliga matches
- 137,765 event records
- 118,607 StatsBomb 360 records
- lineup data for each match
- non-empty 360 data for all 34 matches

## Main Data Structures

### Matches
Match metadata includes:
- match ID
- date
- teams
- score
- match week
- venue
- referee

### Events
Event data contains football actions including:
- passes
- carries
- shots
- pressures
- ball recoveries
- duels
- interceptions
- defensive actions

Event types have different nested schemas. For example, Pass events
contain recipient, length, angle, end location, and outcome information,
while Shot events contain xG, shot outcome, technique, body part, and
shot freeze-frame information.

### Lineups
Lineup files contain team and player metadata, including player position
information.

### StatsBomb 360
360 files link spatial freeze-frame information to event data using
event_uuid.

360 records contain:
- freeze-frame player locations
- teammate/opponent information
- visible-area geometry

360 coverage is extensive but does not exist for every event.

## Conclusion

The dataset provides sufficient event and spatial information for
season-level analysis of Bayer Leverkusen's passing, shooting, ball
progression, defensive actions, player roles, passing networks, heatmaps,
and advanced 360 spatial analysis.
