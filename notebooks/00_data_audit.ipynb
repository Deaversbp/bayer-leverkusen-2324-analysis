```python
def get_leverkusen_result(row):
    if row["home_team"] == "Bayer Leverkusen":
        leverkusen_score = row["home_score"]
        opponent_score = row["away_score"]
    else:
        leverkusen_score = row["away_score"]
        opponent_score = row["home_score"]

    if leverkusen_score > opponent_score:
        return "Win"
    elif leverkusen_score < opponent_score:
        return "Loss"
    else:
        return "Draw"
```

```python
match_summary_df["result"] = match_summary_df.apply(
    get_leverkusen_result,
    axis=1
)
```

```python
match_summary_df["opponent"] = match_summary_df.apply(
    lambda row: (
        row["away_team"]
        if row["home_team"] == "Bayer Leverkusen"
        else row["home_team"]
    ),
    axis=1,
)
```

```python
match_summary_df["venue"] = match_summary_df["home_team"].apply(
    lambda team: "Home" if team == "Bayer Leverkusen" else "Away"
)
```

```python
match_summary_df["goals_for"] = match_summary_df.apply(
    lambda row: (
        row["home_score"]
        if row["home_team"] == "Bayer Leverkusen"
        else row["away_score"]
    ),
    axis=1,
)

match_summary_df["goals_against"] = match_summary_df.apply(
    lambda row: (
        row["away_score"]
        if row["home_team"] == "Bayer Leverkusen"
        else row["home_score"]
    ),
    axis=1,
)
```

```python
match_summary_df["goal_difference"] = (
    match_summary_df["goals_for"]
    - match_summary_df["goals_against"]
)
```
